```python

x = int(input("Enter the value of x: "))
# x is the variable to match
match x:
    # if x is 0
    case 0:
        print("x is zero")
    # case with if-condition
    case 4:
        print("case is 4")

    case _ if x!=90:
        print(x, "is not 90")
    case _ if x!=80:
        print(x, "is not 80")
    case _:
        print(x)
```

```python
x = int(input("Enter the value of x: "))
print("the value of x is: ",x)
match x:
    case _ if x > 50:
        print("x is greater than 50")
    case 50:
        print("x is equal to 50")
    case _:
        print("x is smaller than 50")
```
