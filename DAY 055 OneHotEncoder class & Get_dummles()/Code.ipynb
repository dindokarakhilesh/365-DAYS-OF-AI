# Using Pandas

```python
df = pd.get_dummies(
    data=df,
    columns=['fuel', 'seller_type', 'transmission', 'owner'],
    dtype=int,
    drop_first=True
)
```

# With Sklearn OneHotEcoder¶
When sparse=True, the encoded output is returned as a sparse matrix, while sparse=False returns a dense matrix.

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(
    drop='first',
    dtype=int,
    sparse_output=False
)
```

```python
df_trf = ohe.fit_transform(
    df[['fuel', 'seller_type', 'transmission', 'owner']]
)
```

```python
df_trf
```

```python
cols = ohe.get_feature_names_out()
```

```python
cols
```
