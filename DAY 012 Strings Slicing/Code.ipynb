```python
fruit = "Mango"
mangoLen = len(fruit)
print(mangoLen)
```

```python
print(fruit[0:4]) # including 0 but not 4
print(fruit[1:4]) # including 1 but not 4
```

```python
print(fruit[:5])  
```

```python
print(fruit[0:-3])
print(fruit[:len(fruit)-3])
```

```python
print(fruit[-1:len(fruit) - 3])
print(fruit[-3:-1])
```
