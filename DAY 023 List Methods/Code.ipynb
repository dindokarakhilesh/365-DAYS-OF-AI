```python
l = [11, 45, 1, 2, 6, 4, 1, 1]
print(l)
```

```python
l.append(111)
print(l)
```

```python
l.sort()
print(l)
```

```python
l.sort(reverse=True)
print(l)
```

```python
l.reverse()
print(l)
```

```python
print(l.index(1))
```

```python
print(l.count(1))
```

```python
m = l
m[0] = 0
print(m)
print(l)
```

```python
m = l.copy()
m[0] = 0
print(m)
print(l)
```

```python
l.insert(3, 899)
print(l)
```

```python
m = [900, 1000, 1100]
k = l + m
print(k)
```
