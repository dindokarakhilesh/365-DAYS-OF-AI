```python
# Default arguments:

def name(fname, mname = "Jhon", lname = "Whatson"):
    print("Hello,", fname, mname, lname)

name("Amy")
```

```python
def average(a = 1, b = 1):
    print("The average is: ",(a+b)/2)
```

```python
average()
```

```python
average(2,3)
```

```python
# Keyword arguments:

def name2(fname, mname, lname):
    print("Hello,", fname, mname, lname)

name2(mname = "Peter", lname = "Wesker", fname = "Jade")
```

```python
def num(a,b,c):
    print("a = ",a)
    print("b = ",b)
    print("c = ",c)
```

```python
num(b=10,c=0,a=1)
```

```python
num(c=1,a=2,b=3)
```

```python
# Required arguments:

def name3(fname, mname, lname):
    print("Hello,", fname, mname, lname)

name3("Peter", "Ego")
```

```python
def name3(fname, mname, lname="Quill"):
    print("Hello,", fname, mname, lname)

name3("Peter", "Ego")
```

```python
# Variable-length arguments:
# There are two ways to achieve this:
```

```python
# 1 Arbitrary Arguments:

def name4(*name):
    print("Hello,", name[2], name[1], name[0])

name4("James", "Buchanan", "Barnes")
```

```python
# 2 Keyword Arbitrary Arguments:

def name5(**name):
    print("Hello,", name["fname"], name["mname"], name["lname"])

name5(mname = "Buchanan", lname = "Barnes", fname = "James")
```

```python
# return Statement:

def name6(fname, mname, lname):
    return "Hello, " + fname + " " + mname + " " + lname

print(name6("James", "Buchanan", "Barnes"))
```
