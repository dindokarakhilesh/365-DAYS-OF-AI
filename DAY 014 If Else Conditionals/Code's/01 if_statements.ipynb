```python
a = int(input("Enter your age: "))
print("Your age is:", a)
```

```python
# Conditional operators 
# >, <, >=, <=, ==, !=
print(a>18)
print(a<=18)
print(a==18)
print(a!=18)
```

```python
if(a>18):
  print("You can drive")
  print("Yes")
else:
  print("You cannot drive")
  print("No")
print("Yay!")
```
