```python
tup = (1, 2, 76, 342, 32, "green", True)
print(tup)
print(type(tup))
print(len(tup))
```

```python
# Positive Indexing:

print(tup[0])
print(tup[1])
print(tup[2])
print(tup[3])
print(tup[5])
print(tup[6])
```

```python
# Negative Indexing:

print(tup[-1])  # Similar to print(tup[len(tup) - 1])
print(tup[-2])
print(tup[-3])
print(tup[-4])
print(tup[-5])
print(tup[-6])
```

```python
# Check for item:

if  342 in tup:
  print("Yes 342 is present in this tuple")
else:
  print("No 342 is not present in this tuple")
```

```python
if  3421 in tup:
  print("Yes 3421 is present in this tuple")
else:
  print("No 3421 is not present in this tuple")
```

```python
# Printing elements within a particular range:

print(tup[2:6])     #using positive indexes
print(tup[-7:-2])   #using negative indexes
```

```python
# Printing all element from a given index till the end:

print(tup[4:])      #using positive indexes
print(tup[-4:])     #using negative indexes
```

```python
# printing all elements from start to a given index:

print(tup[:6])      #using positive indexes
print(tup[:-3])     #using negative indexes
```

```python
# Print alternate values:

print(tup[::2])     #using positive indexes
print(tup[-8:-1:2]) #using negative indexes
```

```python
# printing every 3rd consecutive withing given range:

print(tup[1:7:3])
```
