# Confusion Matrix in Machine Learning

## Classification Report | Precision | Recall | Scikit-Learn

This notebook explains the **Confusion Matrix** and important classification metrics using **Scikit-Learn**.

### Topics Covered
- Confusion Matrix
- True Positive (TP)
- True Negative (TN)
- False Positive (FP)
- False Negative (FN)
- Accuracy
- Precision
- Recall
- F1-Score
- Classification Report


## 1. Import Libraries

We will use `scikit-learn` for calculating classification metrics and `matplotlib` for visualization.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

## 2. Sample Classification Data

`y_true` contains the actual values and `y_pred` contains the values predicted by the ML model.

```python
y_true = [1, 1, 1, 1, 0, 0, 0, 0, 1, 0]
y_pred = [1, 1, 0, 1, 0, 1, 0, 0, 1, 0]

print("Actual values :", y_true)
print("Predicted values:", y_pred)
```

## 3. Confusion Matrix

A confusion matrix shows how many predictions were correct and incorrect.

For binary classification:

|                | Predicted 0 | Predicted 1 |
|----------------|-------------:|-------------:|
| **Actual 0**   | TN           | FP           |
| **Actual 1**   | FN           | TP           |

- **TP (True Positive):** Actual = 1, Predicted = 1
- **TN (True Negative):** Actual = 0, Predicted = 0
- **FP (False Positive):** Actual = 0, Predicted = 1
- **FN (False Negative):** Actual = 1, Predicted = 0


```python
cm = confusion_matrix(y_true, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[0, 1]
)

disp.plot()
plt.title("Confusion Matrix")
plt.show()
```

## 4. Accuracy

**Accuracy** tells us the percentage of total predictions that are correct.

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$


```python
accuracy = accuracy_score(y_true, y_pred)
print("Accuracy:", accuracy)
```

## 5. Precision

**Precision** tells us, out of all samples predicted as positive, how many were actually positive.

$$
Precision = \frac{TP}{TP + FP}
$$

High precision means fewer **False Positives**.

```python
precision = precision_score(y_true, y_pred)
print("Precision:", precision)
```

## 6. Recall

**Recall** tells us, out of all actual positive samples, how many were correctly identified.

$$
Recall = \frac{TP}{TP + FN}
$$

High recall means fewer **False Negatives**.

```python
recall = recall_score(y_true, y_pred)
print("Recall:", recall)
```

## 7. F1-Score

The **F1-Score** is the harmonic mean of Precision and Recall.

$$
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

It is useful when we want a balance between precision and recall.

```python
f1 = f1_score(y_true, y_pred)
print("F1-Score:", f1)
```

## 8. Classification Report

Scikit-Learn's `classification_report()` gives **precision, recall, F1-score, and support** for each class.

```python
report = classification_report(y_true, y_pred)
print(report)
```

## 9. Calculate All Main Metrics

```python
print(f"Accuracy :  {accuracy_score(y_true, y_pred):.2f}")
print(f"Precision:  {precision_score(y_true, y_pred):.2f}")
print(f"Recall   :  {recall_score(y_true, y_pred):.2f}")
print(f"F1-Score :  {f1_score(y_true, y_pred):.2f}")
```

## 10. Quick Summary

| Metric | Main Meaning |
|---|---|
| **Accuracy** | Overall correct predictions |
| **Precision** | Correct positive predictions out of predicted positives |
| **Recall** | Correct positive predictions out of actual positives |
| **F1-Score** | Balance between Precision and Recall |
| **Confusion Matrix** | Counts TP, TN, FP and FN |
| **Classification Report** | Complete classification metrics for each class |

### Conclusion

The **Confusion Matrix** helps us understand the types of correct and incorrect predictions made by a classification model. **Precision, Recall, F1-Score, and Classification Report** provide more detailed information about model performance than accuracy alone.
