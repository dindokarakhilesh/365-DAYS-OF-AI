```python
def factorial(num): 
    if (num == 1 or num == 0):
        return 1
    else:
        return (num * factorial(num - 1)) 
  
# Driver Code 
num = 7; 
print("Number: ",num)
print("Factorial: ",factorial(num))
```

```python
# Quick Quiz: Write a program to print the Fibonacci sequence:


def f(n):
    if (n == 0):
        return 0
    elif (n == 1):
        return 1
    else:
        return f(n-1) + f(n-2)
```

```python
n = int(input("Enter a value: "))
print("The value of n is:",n)
print("The Fibonacci sequence of f(",n,") is: ",f(n))
```
